# Generator of Address for Kleros V2 Airdrop
This notebook will generate the list of addresses that met the requirements to receive the PNK airdrop for beign a V2 testnet and a registered PoH human.

More details in this blog post: https://blog.kleros.io/kleros-2-0-juror-101/

Requirements:
1. V2 Testnet user before October 2024
2. PoH registerd human (with same wallet than in point 1), no need to be active at today (29th Nov 2024)

```python
import csv
from typing import Set


def set_to_csv(data: Set, filename: str):
    with open('lists/' + filename, 'w', newline='\n') as file:
        wr = csv.writer(file, quoting=csv.QUOTE_NONE, lineterminator=',\n')
        for item in sorted(data):
            wr.writerow([item])
```

# Airdrop Whitelist

## Coherent Jurors in Testnet
In this section we are going to get the jurors who where coherent at least once

```python
# Using an old list becuase arbitrum-goerli can't be query anymore
with open("lists/All V2 testnet users up to Oct 4th.csv", "r") as file:
    reader = csv.reader(file)
    testnet_jurors = set()
    for row in reader:
        testnet_jurors.add(row[0].lower())
print(f"There are {len(testnet_jurors)} unique juror addresses in the testnet set")
```

## PoH users at snapshot date
Here we check if the jurors in testnet has a valid proof of humanity profile at the snapshot date.

```python
from app.utils.subgraph import PoHV2Subgraph
poh_mainnet = PoHV2Subgraph('mainnet')
profiles_mainnet = poh_mainnet.getRegisteredProfiles()
poh_gnosis = PoHV2Subgraph("gnosis")
profiles_gnosis = poh_gnosis.getRegisteredProfiles()

print("There are {} profiles on mainnet".format(len(profiles_mainnet)))
print("There are {} profiles on gnosis".format(len(profiles_gnosis)))

```

```python
profiles = set(profiles_mainnet + profiles_gnosis)
print(
    f"There are {len(profiles)} unique profile addresses in the proof of humanity set"
)
```

```python
set_to_csv(profiles, "poh-submissions-nov29-2024.csv")
```

## Intersection PoH and Testnet users

```python
whitelist = testnet_jurors.intersection(profiles)
print(f"There are {len(whitelist)} addresses that match all the requirements in the set")
print(whitelist)
set_to_csv(whitelist, 'whitelist_airdrop_poh_testnet.csv', )
```
